```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/Users/pratikvilasgadhe/Desktop/Programming/AI_ML/practical/Housing.csv")
df.info()
```

```python
from sklearn.preprocessing import OneHotEncoder

categorical_columns = df.select_dtypes(include = ['object']).columns
categorical_columns
encoder = OneHotEncoder()

encoded_data = encoder.fit_transform(df[categorical_columns])
encoded_data

encoded_df = pd.DataFrame(encoded_data.toarray(), columns=encoder.get_feature_names_out(categorical_columns))
encoded_df

final_df = pd.concat([df.drop(columns=categorical_columns),encoded_df],axis=1)
final_df
```

```python
x = final_df.drop(columns='price')  #input feature
y = final_df['price'] #output label
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state = 42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
y_predict = model.predict(x_train)
```

```python
from sklearn.metrics import r2_score
r2_score(y_train,y_predict)
```
