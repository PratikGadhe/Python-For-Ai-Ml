```python
import numpy as np 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# 1. Load the Housing dataset
df = pd.read_csv(r"/Users/pratikvilasgadhe/Desktop/Programming/AI_ML/practical/Housing.csv")
df.info()
df.describe()
df.head()
```

```python
# simple linear regression
X = df[['area']].values       #independent variable or input variable
Y = df['price'].values      #dependent variable or output variable
```

```python
#  split data into training and testing
from sklearn.model_selection import train_test_split
x_train,x_text,y_train,y_test = train_test_split(X,Y,test_size=0.2,random_state = 42)
print(X.shape)
```

```python
from sklearn.linear_model import LinearRegression
lr_model = LinearRegression()
lr_model.fit(X, Y)
y_pred = lr_model.predict(X)

```

```python
# metrices 
from sklearn.metrics import mean_squared_error, r2_score
import seaborn as sns
rmse_simple = np.sqrt(mean_squared_error(Y, y_pred))
r2_simple = r2_score(Y, y_pred)
print("--- Simple Linear Regression (Scikit-Learn) ---")
print(f"Intercept: {lr_model.intercept_:.2f} | Slope: {lr_model.coef_[0]:.2f}")
print(f"RMSE: {rmse_simple:.2f} | R²: {r2_simple:.4f}\n")
```
